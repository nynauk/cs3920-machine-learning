```python
# In[1]
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

cancer = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    cancer.data, cancer.target, random_state=42
)
print(X_train.shape)
print(X_test.shape)
```

```python
# In[2]
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
```

```python
# In[3]
scaler.fit(X_train)
```

```python
# In[4]
# transform data
X_train_scaled = scaler.transform(X_train)
# print dataset properties before and after scaling
print(X_train_scaled.shape)
print(X_train.min(axis=0))
print(X_train.max(axis=0))
print(X_train_scaled.min(axis=0))
print(X_train_scaled.max(axis=0))
```

```python
# In[5]
# transform test data
X_test_scaled = scaler.transform(X_test)
# print test data properties after scaling
print(X_test_scaled.min(axis=0))
print(X_test_scaled.max(axis=0))
```

```python
# In[6]
from sklearn.svm import SVC
svm = SVC(C=1)
svm.fit(X_train, y_train)
print(svm.score(X_test, y_test))
```

```python
# In[7]
X_test_scaled = scaler.transform(X_test)
# learning an SVM on the scaled training data:
svm.fit(X_train_scaled, y_train)
# scoring on the scaled test set:
print(svm.score(X_test_scaled, y_test))
```

```python
# In[8]
# naive grid search implementation
from sklearn.model_selection import train_test_split  # safe re-import in notebook cells

X_train, X_test, y_train, y_test = train_test_split(
    cancer.data, cancer.target, random_state=42
)
best_score = 0
for gamma in [0.001, 0.01, 0.1, 1, 10, 100]:
    for C in [0.001, 0.01, 0.1, 1, 10, 100]:
        # for each combination of parameters, train an SVM
        svm = SVC(gamma=gamma, C=C)
        svm.fit(X_train, y_train)
        # evaluate the SVM on the test set
        score = svm.score(X_test, y_test)
        # if we got a better score, store the score and parameters
        if score > best_score:
            best_score = score
            best_C = C
            best_gamma = gamma
print("Best score:", best_score)
print("Best parameters C and gamma:", best_C, best_gamma)
```

```python
# In[9]
# split data into train and test sets
X_train, X_test, y_train, y_test = train_test_split(
    cancer.data, cancer.target, random_state=42
)
# split train set into train set proper and validation set
X_train_pr, X_valid, y_train_pr, y_valid = train_test_split(
    X_train, y_train, random_state=42
)
print("Sizes of train_pr, valid, and test sets:",
      X_train_pr.shape[0], X_valid.shape[0], X_test.shape[0])

best_score = 0
for gamma in [0.001, 0.01, 0.1, 1, 10, 100]:
    for C in [0.001, 0.01, 0.1, 1, 10, 100]:
        # for each combination of parameters, train an SVM
        svm = SVC(gamma=gamma, C=C)
        svm.fit(X_train_pr, y_train_pr)
        # evaluate the SVM on the validation set
        score = svm.score(X_valid, y_valid)
        # if we got a better score, store the score and parameters
        if score > best_score:
            best_score = score
            best_C = C
            best_gamma = gamma

# rebuild a model on the full training set, and evaluate it on the test set
svm = SVC(C=best_C, gamma=best_gamma)
svm.fit(X_train, y_train)
test_score = svm.score(X_test, y_test)
print("Best score on validation set:", best_score)
print("Best parameters C and gamma:", best_C, best_gamma)
print("Test set score with best parameters:", test_score)
```

```python
# In[10]
import numpy as np
from sklearn.model_selection import cross_val_score

best_score = 0
for gamma in [0.001, 0.01, 0.1, 1, 10, 100]:
    for C in [0.001, 0.01, 0.1, 1, 10, 100]:
        # for each combination of parameters, train an SVC
        svm = SVC(gamma=gamma, C=C)
        # perform cross-validation
        scores = cross_val_score(svm, X_train, y_train, cv=5)
        # compute mean cross-validation accuracy
        score = np.mean(scores)
        # if we got a better score, store the score and parameters
        if score > best_score:
            best_score = score
            best_C = C
            best_gamma = gamma

# rebuild a model on the full training set
svm = SVC(C=best_C, gamma=best_gamma)
svm.fit(X_train, y_train)
test_score = svm.score(X_test, y_test)
print("Best CV score:", best_score)
print("Best parameters C and gamma:", best_C, best_gamma)
print("Test set score with best parameters:", test_score)
```

### Task 1

```python
# Task 1 — Load the ORIGINAL tab-delimited dataset (NOT sklearn's)
import pandas as pd

df = pd.read_csv("diabetes.data", delimiter="\t")
print(df.head(3))
print(df.shape)
```

### Task 2
The training set contains 331 samples, and the test set contains 111 samples (a 75/25 split).

```python
# Task 2 — Split into train/test with a fixed random_state
from sklearn.model_selection import train_test_split

random_state = 1208  # fixed seed
X = df.drop(columns=["Y"])
y = df["Y"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=random_state
)
X_train.shape, X_test.shape
```

### Task 3
The model explains about 54% of the variance in training data and 38% in test data.
This suggests moderate underfitting due to strong regularization and unscaled features.

```python
# Task 3 — Lasso (default params) on RAW data: report train/test R² and #non-zero features
import numpy as np
from sklearn.linear_model import Lasso

lasso_default = Lasso()
lasso_default.fit(X_train, y_train)

r2_train_default = lasso_default.score(X_train, y_train)
r2_test_default = lasso_default.score(X_test, y_test)
nonzero_default = np.count_nonzero(lasso_default.coef_)

print("Task 3 — Lasso (raw)")
print("Training R^2:", round(r2_train_default, 3))
print("Test R^2:", round(r2_test_default, 3))
print("Non-zero features:", int(nonzero_default))

```

### Task 4
A StandardScaler was fitted only on the training data to prevent data snooping.
Both training and test sets were transformed using this fitted scaler.
This ensures that the scaling process does not leak information from the test set into training.

```python
# Task 4 — Normalize with StandardScaler fitted ONLY on TRAIN (no snooping)
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler().fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled  = scaler.transform(X_test)

X_train_scaled[:2], X_test_scaled[:2]
```

### Task 5
Normalization slightly improved model performance and reduced the number of active features.
This shows that scaling helps regularization work more effectively, improving generalization.

```python
# Task 5 — Lasso (default params) on SCALED data
lasso_scaled = Lasso()
lasso_scaled.fit(X_train_scaled, y_train)

r2_train_scaled = lasso_scaled.score(X_train_scaled, y_train)
r2_test_scaled = lasso_scaled.score(X_test_scaled, y_test)
nonzero_scaled = np.count_nonzero(lasso_scaled.coef_)

print("Task 5 — Lasso (scaled)")
print("Training R^2:", round(r2_train_scaled, 3))
print("Test R^2:", round(r2_test_scaled, 3))
print("Non-zero features:", int(nonzero_scaled))
```

### Task 6
Cross-validation selected an optimal α ≈ 0.81, balancing bias and variance.
The R² values are consistent with the scaled Lasso model, indicating that the model generalizes well.
Only 8 features are retained, meaning Lasso performs effective feature selection by setting weak coefficients to zero.

```python
# Task 6 — Choose alpha via 5-fold CV on the NORMALIZED TRAIN set; then report R²
from sklearn.linear_model import LassoCV

lasso_cv = LassoCV(cv=5, random_state=random_state)
lasso_cv.fit(X_train_scaled, y_train)

best_alpha = lasso_cv.alpha_
r2_train_cv = lasso_cv.score(X_train_scaled, y_train)
r2_test_cv  = lasso_cv.score(X_test_scaled, y_test)
nonzero_cv  = np.count_nonzero(lasso_cv.coef_)

print("Task 6 — LassoCV on scaled data")
print("Best alpha:", round(best_alpha, 6))
print("Training R^2:", round(r2_train_cv, 3))
print("Test R^2:", round(r2_test_cv, 3))
print("Non-zero features:", int(nonzero_cv))
```
